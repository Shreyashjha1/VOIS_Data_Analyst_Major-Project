## Phase 4: Seasonal Performance Analysis ✅
- Compared yield & production across seasons
- Visualized seasonal trends


```python
import seaborn as sns
import matplotlib.pyplot as plt

seasonal_summary = df.groupby("Season")[["Yield_Tonnes_Ha", "Production_Tonnes"]].mean().reset_index()

plt.figure(figsize=(8,5))
sns.barplot(data=seasonal_summary, x="Season", y="Yield_Tonnes_Ha", palette="viridis")
plt.title("Average Yield (Tonnes/Ha) by Season")
plt.show()

plt.figure(figsize=(8,5))
sns.barplot(data=seasonal_summary, x="Season", y="Production_Tonnes", palette="magma")
plt.title("Average Production (Tonnes) by Season")
plt.show()

plt.figure(figsize=(8,5))
sns.lineplot(data=seasonal_summary, x="Season", y="Yield_Tonnes_Ha", marker="o", label="Yield_Tonnes_Ha")
sns.lineplot(data=seasonal_summary, x="Season", y="Production_Tonnes", marker="o", label="Production_Tonnes")
plt.title("Seasonal Yield vs Production Trends")
plt.legend()
plt.show()
```

## Phase 5: Relationship & Comparative Analysis ✅
- Correlation heatmap created
- Scatterplots for Rainfall, Temperature, Humidity vs Yield/Production
- Comparative analysis completed


```python
import seaborn as sns
import matplotlib.pyplot as plt

corr = df.corr(numeric_only=True)
plt.figure(figsize=(12,8))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()

plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Rainfall_mm", y="Yield_Tonnes_Ha", hue="Season", palette="viridis")
plt.title("Rainfall vs Yield by Season")
plt.show()

plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Avg_Temperature_C", y="Production_Tonnes", hue="Season", palette="magma")
plt.title("Temperature vs Production by Season")
plt.show()

plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Humidity_pct", y="Yield_Tonnes_Ha", hue="Season", palette="plasma")
plt.title("Humidity vs Yield by Season")
plt.show()
```

## Phase 6: Key Insights & Interpretation ✅
- Rainfall strongly correlated with Yield_Tonnes_Ha
- Higher Avg_Temperature_C linked to lower Production_Tonnes
- Humidity_pct shows moderate impact on yield
- Certain seasons consistently outperform others in both yield and production
- Irrigation_Method and Fertilizer_kg_ha influence efficiency


```python
import seaborn as sns
import matplotlib.pyplot as plt

corr = df[["Rainfall_mm","Avg_Temperature_C","Humidity_pct","Yield_Tonnes_Ha","Production_Tonnes"]].corr()
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Key Variable Correlations")
plt.show()

plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Rainfall_mm", y="Yield_Tonnes_Ha", palette="viridis")
plt.title("Rainfall vs Yield")
plt.show()

plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Avg_Temperature_C", y="Production_Tonnes", palette="magma")
plt.title("Temperature vs Production")
plt.show()

plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Humidity_pct", y="Yield_Tonnes_Ha", palette="plasma")
plt.title("Humidity vs Yield")
plt.show()
```
